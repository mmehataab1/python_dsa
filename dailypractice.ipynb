```python
import tkinter as tk
import random
import time

arr = []

def generate():
    global arr
    arr = [random.randint(20, 250) for _ in range(25)]
    draw()

def draw(highlight=[]):
    canvas.delete("all")
    for i, value in enumerate(arr):
        x1 = i * 20
        y1 = 300 - value
        x2 = x1 + 18
        y2 = 300
        canvas.create_rectangle(
            x1, y1, x2, y2,
            fill="red" if i in highlight else "blue"
        )
    root.update()

def bubble_sort():
    n = len(arr)

    for i in range(n):
        for j in range(n - i - 1):
            draw([j, j + 1])
            time.sleep(0.05)

            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

    draw()

def selection_sort():
    n = len(arr)

    for i in range(n):
        minimum = i

        for j in range(i + 1, n):
            draw([minimum, j])
            time.sleep(0.05)

            if arr[j] < arr[minimum]:
                minimum = j

        arr[i], arr[minimum] = arr[minimum], arr[i]

    draw()

def insertion_sort():
    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            draw([j, j + 1])
            time.sleep(0.05)

            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    draw()


root = tk.Tk()
root.title("Sorting Visualizer")
root.geometry("520x380")

canvas = tk.Canvas(root, width=500, height=310)
canvas.pack()

tk.Button(root, text="Generate Array", command=generate).pack(side="left", padx=10)
tk.Button(root, text="Bubble Sort", command=bubble_sort).pack(side="left", padx=5)
tk.Button(root, text="Selection Sort", command=selection_sort).pack(side="left", padx=5)
tk.Button(root, text="Insertion Sort", command=insertion_sort).pack(side="left", padx=5)

generate()

root.mainloop()
```
