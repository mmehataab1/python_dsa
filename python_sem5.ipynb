```python
#transaction system
print("Transaction system")
balance =0
transactions= []

while True:
  print("1.Deposit")
  print("2.Withdraw")
  print("3.Check balance")
  print("4.Transaction history")
  print("5.Exit")

  choice= int(input("Enter your choice:"))

  if choice == 1:
    amt= int(input("Enter the amount to deposit:"))
    balance += amt
    transactions.append("Deposited"+ str(amt))
    print("amount deposited")

  elif choice == 2:
    amt= int(input("Enter the amount to withdraw:"))
    if amt <= balance:
      balance -= amt
      transactions.append("Withdrawn"+ str(amt))
      print("amount withdrawn")
    else:
      print("insufficient balance")

  elif choice == 3:
    print("current balance:",balance)
  elif choice == 4:
    print("Transaction history:")
    for t in transactions:
      print(t)
  elif choice == 5:
    print("Thank you for using the transaction system")
    break
  else:
    print("invalid choice")
```

```python
#implementing stack using list
class stack:
  def __init__(self):
    self.items =[]

  def push(self,x):
    self.items.append(x)

  def pop(self,x):
    if self.is_empty():
      raise IndexError('pop from empty list')
    return
    self.items.pop(x)

  def peek(self,x):
    if self.is_empty():
      raise IndexError('peek from empty list')
    return
    self.items[-1]

  def is_empty(self):
    return len(self.items) == 0

  def size(self):
    return len(self.items)

s= stack()
print(type(s))
s.push(20)
s.push(30)
s.push(40)
s.push(50)
print(s.pop(50))
print(s.peek(40))
print(s.size())
print(s.is_empty())
```

```python
#queue implimentation using list
class Queue:
  def __init__(self):
    self.items=[]

  def enqueue(self,x):
    self.items.append(x)

  def dequeue(self):
    if self.is_empty():
      raise IndexError('dequeue from empty list')
    return
    self.items.pop(0)

  def front(self):
    if self.is_empty():
      raise IndexError('front from empty list')
    return
    self.items[0]

  def is_empty(self):
    return len(self.items) == 0

q = Queue()
q.enqueue(7)
q.enqueue(10)
q.enqueue(17)
q.dequeue()
q.front()
print()
```

```python
#In the Gregorian calendar, three conditions are used to identify leap years:The year can be evenly divided by 4, is a leap year, unless:The year can be evenly divided by 100, it is NOT a leap year, unless:
#The year is also evenly divisible by 400. Then it is a leap year.This means that in the Gregorian calendar, the years 2000 and 2400 are leap years, while 1800, 1900, 2100, 2200, 2300 and 2500 are NOT leap years
def is_leap(year):
  leap = False
  if year % 400 == 0:
    leap = True
  elif year % 100 ==0 and year % 4 == 0:
    leap = True
  else:
    return leap

year = int(input())
print(is_leap(year))
```

```python
#two sum basic one (brute force approach)
def two_sum(num,target):
  n=len(num)
  for i in range(n):
    for j in range(i+1,n):
      if num[i] + num [j]== target:
        return [i,j]
    return []

num= [2,7,11,15]
two_sum(num,17)
```

```python
#two sum hashmap approach
class solution(object):
  def two_sum(self,nums,target):
    my_map={}
    for i,num in enumerate(nums):
      compliment=target-num
      if compliment in my_map:
        return [my_map[compliment],i]
      my_map[num]=i

nums=[2,7,11,15]
target=9
result = two_sum(nums,target)
print(result)


```

```python
def max_subarray(nums):
# Initialize both tracking variables with the first element
    max_so_far = nums[0]
    current_max = nums[0]

# Loop through the rest of the array starting from index 1
    for x in nums[1:]:
# Decide whether to add the current element to the existing subarray
# or start a new subarray from this element.
        current_max = max(x, current_max + x)

# Update the overall maximum found so far
        max_so_far = max(max_so_far, current_max)

    return max_so_far

# Example usage:
nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print("Maximum Subarray Sum:", max_subarray(nums))
```

```python
#kadane algorithm
def kadane(arr):
  max_sum=arr[0]
  current_sum = arr[0]

  for i in range (1,len(arr)):

    current_sum = max(arr[i], current_sum + arr[i])
    max_sum = max(current_sum,max_sum)

  return max_sum

arr = [7,-5,3,6,-2,7]
kadane(arr)

```

```python
from abc import ABC, abstractmethod
class Employee(ABC):
    def __init__(self, name, emp_id):
        self.name = name
        self.emp_id = emp_id
    def calculate_salary(self):
        pass
    def display_details(self):
        pass
class PermanentEmployee(Employee):
    def __init__(self, name, emp_id, basic_pay, bonus):
        super().__init__(name, emp_id)
        self.basic_pay, self.bonus = basic_pay, bonus
    def calculate_salary(self):
        return self.basic_pay + self.bonus
    def display_details(self):
        print(
            f"Permanent | ID: {self.emp_id} | Name: {self.name} | Salary: ${self.calculate_salary()}"
)
class ContractEmployee(Employee):
    def __init__(self, name, emp_id, hourly_rate, hours_worked):
        super().__init__(name, emp_id)
        self.hourly_rate, self.hours_worked = hourly_rate, hours_worked

    def calculate_salary(self):
        return self.hourly_rate * self.hours_worked

    def display_details(self):
        print(
            f"Contract  | ID: {self.emp_id} | Name: {self.name} | Salary: ${self.calculate_salary()}"
        )

emp1 = PermanentEmployee("Sarah", "P101", 5000, 1200)
emp2 = ContractEmployee("David", "C204", 40, 160)

emp1.display_details()
emp2.display_details()
```

```python
def maxDepth (root):
  if root is None:
    return 0

    leftDepth = maxDepth(root.left)
    rightDepth = maxDepth(root.right)

    return max(leftDepth,rightDepth) + 1
```

```python
def height(root):
  if root is None:
    return -1
  else:
    left = height(root.left)
    right = height(root.right)
    return max(left,right) + 1
```

```python
from collections import deque
graph = {
    0: [1,2,3],
    1: [0,3,4],
    2: [0,3],
    3: [0,1,2,4],
    4: [1,3]
}
print(graph)
#breadth first search
def bfs(graph,start):
  queue = deque([start])
  visited = set([start])
  while queue:
    node = queue.popleft()
    print(node)
    for neighbour in graph[node]:
      if neighbour not in visited:
        visited.add(neighbour)
        queue.append(neighbour)

bfs(graph,1)
```

```python
class stack:
  def __init__(self):
    self.items =[]

  def push(self,x):
    self.items.append(x)

  def pop(self,x):
    if self.is_empty():
      raise IndexError('pop from empty list')
    return
    self.items.pop(x)

  def peek(self,x):
    if self.is_empty():
      raise IndexError('peek from empty list')
    return
    self.items[-1]

  def is_empty(self):
    return len(self.items) == 0

  def size(self):
    return len(self.items)

s= stack()
print(type(s))
s.push(20)
s.push(30)
s.push(40)
s.push(50)
print(s.pop(50))
print(s.peek(40))
print(s.size())
print(s.is_empty())
```

```python
#queue implimentation using list
class Queue:
  def __init__(self):
    self.items=[]

  def enqueue(self,x):
    self.items.append(x)

  def dequeue(self):
    if self.is_empty():
      raise IndexError('dequeue from empty list')
    return
    self.items.pop(0)

  def front(self):
    if self.is_empty():
      raise IndexError('front from empty list')
    return
    self.items[0]

  def is_empty(self):
    return len(self.items) == 0

q = Queue()
q.enqueue(7)
q.enqueue(10)
q.enqueue(17)
q.dequeue()
q.front()
print()
```
